```python
"""
Will Buesnel, Aug 26

Just going to produce some toy figures to explain bayesian inference.

just going to produce 3 normal distributions on the same figure
"""

from scipy.stats import norm
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

import matplotlib as mpl

mpl.rcParams["axes.labelsize"] = 18      # x and y axis labels
mpl.rcParams["xtick.labelsize"] = 16     # x-axis tick labels
mpl.rcParams["ytick.labelsize"] = 16     # y-axis tick labels
mpl.rcParams["legend.fontsize"] = 16   # legend text
mpl.rcParams["legend.title_fontsize"] = 18
mpl.rcParams["axes.titlesize"] = 20      # plot title
```

```python
mu1, sigma1 = 1.5, 0.2 # mean and standard deviation
mu2, sigma2 = 0, 0.25
mu3, sigma3 = 1.8, 0.3

colours = sns.color_palette("Set2", 3)
# create a range of x values
x = np.linspace(-1, 3, 1000)
# plot via scipy.stats.norm.pdf
plt.figure(figsize=(8, 6))
plt.plot(x, norm.pdf(x, mu1, sigma1), color=colours[0], label=r"posterior distribution")
plt.plot(x, norm.pdf(x, mu2, sigma2), color=colours[1], label=r"prior distribution")
plt.plot(x, norm.pdf(x, mu3, sigma3), color=colours[2], label=r"true distribution")

# fill the area under the curves
plt.fill_between(x, norm.pdf(x, mu1, sigma1), alpha=0.3, color=colours[0])
plt.fill_between(x, norm.pdf(x, mu2, sigma2), alpha=0.3, color=colours[1])
plt.fill_between(x, norm.pdf(x, mu3, sigma3), alpha=0.3, color=colours[2])
plt.title(r"Distributions Before Inference")
plt.xlabel("x")
plt.ylabel("Probability density")
plt.legend()
plt.savefig("bayesian_inference_toy_example.pdf", dpi=300)
```

```python
# do a different example, with a different prior and posterior, that results in a posteriors that is much more similar to the true dsitribution.

mu1, sigma1 = 1.7, 0.3 # mean and standard deviation
mu2, sigma2 = 1.3, 0.25
mu3, sigma3 = 1.8, 0.3

plt.figure(figsize=(8, 6))
plt.plot(x, norm.pdf(x, mu1, sigma1), color=colours[0], label=r"posterior distribution")
plt.plot(x, norm.pdf(x, mu2, sigma2), color=colours[1], label=r"prior distribution")
plt.plot(x, norm.pdf(x, mu3, sigma3), color=colours[2], label=r"true distribution")

# fill the area under the curves
plt.fill_between(x, norm.pdf(x, mu1, sigma1), alpha=0.3, color=colours[0])
plt.fill_between(x, norm.pdf(x, mu2, sigma2), alpha=0.3, color=colours[1])
plt.fill_between(x, norm.pdf(x, mu3, sigma3), alpha=0.3, color=colours[2])
plt.title(r"Distributions After Inference")
plt.xlabel("x")
plt.ylabel("Probability density")
plt.legend()
plt.savefig("bayesian_inference_toy_example_2.pdf", dpi=300)
```
